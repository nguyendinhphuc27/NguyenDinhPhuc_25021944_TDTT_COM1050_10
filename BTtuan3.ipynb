```python
#W3A11
a, b, c = map(float, input('Mời nhập 3 só cần kiểm tra:').split())
if (a-b < c < a+b) and (a-c <b < a+c) and (b-c <a < b+c) :
    if a==b==c:
      print('Ba số trên lập thành một tam giác đều!')
    elif a==b!=c or a==c!=b or b==c!=a:
      print('Ba số trên lập thành một tam giác cân!')
    elif (a== pow(b**2+c**2,0.5)) or (b== pow(a**2+c**2,0.5)) or (c== pow(a**2+b**2,0.5)):
        print('Ba số trên lập thành tam giác vuông!')
    else:
      print('Ba số trên lập thành một tam giác thường!')
else:
    print('Ba số trên không lập thành một tạm giác!')
```
