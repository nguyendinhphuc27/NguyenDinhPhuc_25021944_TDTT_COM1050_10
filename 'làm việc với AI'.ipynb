```python
#W1A1
a=7       #gán biến a với giá trị bằng số nguyên là 7
b=5       #gán biến b với giá trị bằng số nguyên là 5
c=a-b     #gán biến c với giá trị bằng a-b (c=2)
print(c)  #in biến c ra màn hình
```

```python
#W1A2
city='Hà Nội'   #gán biến "city" với giá trị là 1 sâu "hà nội"
year=2025       #gán biến 'year' với giá trị là 1 số nguyên
print('Thành phố:',city,'-Năm:',year,sep=' ')   #in ra màn hình dòng chữ 'Thành phố: Hà Nội-Năm:2025"
```

```python
#W1A3
n=4          
t=0
for i in range(1,n+1):    #cấu trúc câu lệnh vòng lặp với i chạy từ 1 đến giá trị (n+1)-1=n
    t+=1   #sau mỗi 1 vòng lặp tăng giá trị t thành t=t+1
    print(t)
```

```python
#W1A4
num=[91,24,35,37,36,38] #gán biến 'nun' là 1 danh sách với các phần từ ở trong 1 cặp','
for i in num:           # cho biến i chạy lần lượt là các phần tử trong dsach 'num'
    if i%2==0:          #nếu i chia hết cho 2 (i chia 2 dư 0)
        print(i,'là số chẵn',sep=' ')
    else:
        print(i,'là số lẻ',sep=' ')
```

```python
#W1A5
animals=['cat','bear','cow']  #lưu biến 'animals' là 1 dsach
count=0       #gán 'count' với gtri =0
for i in animals:
    count+=1
print('Số phần tử trong danh sách là:',count,sep=' ')
```
