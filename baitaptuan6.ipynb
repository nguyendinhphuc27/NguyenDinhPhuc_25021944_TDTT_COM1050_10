```python
#W6A1
def check(lst) :
    lst_r = []
    for i in range(len(lst)):
        if lst[i] not in lst_r :
            lst_r.append(lst[i])
    return lst_r
lst = list(map(int, input().split()))
print(check(lst))
```

```python
#W6A2
def sums(lst) :
    s = 0
    kq = []
    for i in range(len(lst)) :
        s += lst[i]
        kq.append(s)
    return kq
lst = list(map(int, input().split()))
print(sums(lst))
```

```python
#W6A3
def swap(tup, k) :
    lst = list(i for i in tup)
    kq = []
    while k :
        a_tam = lst[0]
        lst.append(a_tam)
        k -= 1
        lst.remove(lst[0])
    kq = tuple(i for i in lst)
    return kq
tup = tuple(map(int, input().split()))
k = int(input())
print(swap(tup,k))
```

```python
#W6A4
c = list(map(str, input().split()))
print(c)
def check(c) :
    dct = {}
    for i in range(0,len(c), 2): 
        #khi có step bắt buộc phải có đủ start và stop
        key_test = c[i]
        value = c[i+1]
        if key_test in dct :
            dct[key_test].append(value)
        else :
            dct[key_test] = [value]
    return dct
print(check(c))
```

```python
#W6A5
def ktra(lst) -> dict :
    p, n, z = 0, 0, 0
    dct = {}
    for i in range(len(lst)) :
        if lst[i] > 0 :
            p += 1
        elif lst[i] < 0 :
            n += 1
        else :
            z += 1
    dct['positives'] = p
    dct['negatives'] = n
    dct['zeros'] = z
    return dct
lst = list(map(int, input().split()))
print(ktra(lst))
```

```python
#W6A8
def count_(lst) -> dict :
    dic = {}
    for i in range(len(lst)) :
        if lst[i] in dic :
            dic[lst[i]] += 1
        else :
            dic[lst[i]] = 1
    return dic
lst = list(map(str, input().split()))
print(count_(lst))
```

```python
#W6A9
def combine(dic1, dic2) :
    k1, k2 = len(dic1), len(dic2)
    if k1 < k2 :
        for i in dic1.keys() :
            if i in dic2.keys() :
                for k in dic1[i] :
                    dic2[i].append(k)
            else:
                dic2[i] = dic1[i]
        return dic2
    else:
        for i in dic2.keys() :
            if i in dic1.keys() :
                for k in dic2[i] :
                    dic1[i].append(k) 
            else:
                dic1[i] = dic2[i]
        return dic1
dct1, dct2 = {}, {}
while True :
    key = input('Mời nhập key cho 2 dict').split()
    value1 = input('Value cho dict 1').split()
    value2 = input('Value cho dict 2').split()
    dct2[key[1]] = value2
    dct1[key[0]] = value1
    flag = input('Dừng/không dừng: Trả lời dưới dạng Y/N')
    if flag == 'Y' or flag == 'y':
        break
print(dct1)
print(dct2)
print(combine(dct1,dct2))
```

```python
#W6A10
def check(lst, k) -> list :
    kq = []
    for i in range(len(lst)) :
        for x in range(i+1, len(lst)) :
            if lst[i] + lst[x] == k :
                tup = (lst[i], lst[x])
                tup_doi = (lst[x], lst[i])
                if tup not in kq and tup_doi not in kq:
                    kq.append(tup)
    kq.sort()
    return kq
lst = list(map(int, input().split()))
k = int(input())
print(check(lst, k))
```

```python
#W6A11
def decompose(tup):
    lst = list(i for i in tup)
    l1, l2 = [], []
    for i in lst :
        if i %2 == 0 :
            l1.append(i)
        else:
            l2.append(i)
    l1 = tuple(i for i in l1)
    l2 = tuple(i for i in l2)
    return l1, l2
tup = tuple(map(int, input().split()))
print(decompose(tup)[0],'\n', decompose(tup)[1])
```

```python
#W6A12
def find_frequency(lst) :
    lst.sort()
    a_max = 0
    b_min_max = 0
    for i in lst :
        a = 0
        for x in lst :
            if x == i :
                a +=1
        if a-1 > a_max :
            a_max = a -1
            b_min_max = i
    return b_min_max
lst = list(map(int, input().split()))
print(find_frequency(lst))
```

```python
#W6A13
def swap_key_value(dct) :
    result_dct = {}
    for i in dct :
        result_dct[dct[i]] = i 
    return result_dct
dct = {}
while True :
    key = input('Mời nhập key cho  dict')
    value = input('Value cho dict, value là duy nhất kh trùng nhau!!')
    dct[key] = value
    flag = input('Dừng/không dừng: Trả lời dưới dạng Y/N')
    if flag == 'Y' or flag == 'y':
        break
print(swap_key_value(dct))
```

```python
#W6A15
def loc_dic(dct, k) :
    kq = {}
    for i in dct :
        if dct[i] > k :
           kq[i] = dct[i] 
    return kq
dct = {}
while True :
    key = input('Mời nhập key cho dict')
    value = int(input('Value cho dict, value là số nguyên!!'))
    dct[key] = value
    flag = input('Dừng/không dừng: Trả lời dưới dạng Y/N')
    if flag == 'Y' or flag == 'y':
        break
k = int(input())
print(loc_dic(dct, k))
```
