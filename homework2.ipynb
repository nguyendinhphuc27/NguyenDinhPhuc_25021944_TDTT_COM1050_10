```python
#HW2.5
a, b, c =map(float,input('Mời nhập hệ các hệ số a, b, c của phương trình bậc 2:').split())
delta = b**2 -4*a*c
if delta > 0 :
    x1 = (-b+pow(delta,0.5))/(2*a)
    x2 = (-b-pow(delta,0.5))/(2*a)
    print(f'Nghiệm của phương trình bậc 2 y=ax**2+bx+c là :{x1} và {x2}')
elif delta ==0:
    x = (-b+pow(delta,0.5))/(2*a)
    print(f'Nghiệm của phương trình bậc 2 y=ax**2+bx+c là :{x}')
else:
    print(f'Phương trình bậc 2 y={a}x**2+{b}x+{c} vô nghiệm')
```
