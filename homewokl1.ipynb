```python
#HW1.3
''' 
  _hàm kiểm tra ký tự hoa và ký tự thường_
''biến'.isupper()': kiểm tra xem có phải kí tự hoa kh
''biến'.islower()': kiểm tra xem có phải kí tự thường kh
''biến'.lower()':chuyển từ ký tự hoa sang ký tự thường
''biến'.upper()':chuyển từ ký tự thg sang ký tự hoa
'''
a= input('Mời bạn nhập 1 ký tự:')
if a.isupper()== True:
    print(a.lower())
else:
    print(a.upper())
```

```python
#HW1-BT khác 8
def kiem_tra_nam_nhuan(nam):
    if nam %100 == 0 or (nam %4 == 0 and nam %100 != 0):
        print('Năm', nam, 'là năm nhuận',sep=' ')
    else:
        print('Năm', nam, 'không là năm nhuận', sep=' ')
'''
try...except :Dùng để ktra 1 khối lệnh có chạy kh khi nghi ngờ khối lệnh bị lỗi, 
                 nếu đoạn lệnh lỗi thì sẽ chạy sửa đoạn đó mà kh dừng đột ngột
'''
try:
    year=int(input('Mời bạn nhập năm cần kiểm tra:'))
    if year<0:
        print('Mời nhập lại! Năm phải là số dương.')
    else:
        kiem_tra_nam_nhuan(year)
except ValueError:
    print('Năm phải là số nguyên!Mời bạn nhập lại.')
        
```
