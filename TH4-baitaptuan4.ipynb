```python
#W4A8
def find_palindrome(n_input):
    """Tìm số Palindrome (đối xứng) bằng phép toán Đảo ngược và Cộng."""
    
    n = n_input
    steps = 0
    
    # Lặp cho đến khi n là số đối xứng
    while str(n) != str(n)[::-1]: 
        
        # 1. Đảo ngược: Lấy chuỗi của n và đảo ngược nó
        reversed_n = int(str(n)[::-1])
        
        # 2. Cộng: Thực hiện phép toán n = n + reverse(n)
        n += reversed_n
        steps += 1
      
    print(f"Bắt đầu với: {n_input}")
    print(f"Số bước: {steps}")
    print(f"Số Palindrome: {n}")
find_palindrome(84)
```

```python
def collatz_sequence_length(n):
    """
    Tính độ dài của chuỗi Collatz bắt đầu từ n.
    Sử dụng bộ nhớ đệm (memoization) để tối ưu hiệu suất 
    cho các bài toán lớn hơn, mặc dù không bắt buộc cho giới hạn nhỏ.
    """
    # Dictionary để lưu trữ độ dài đã tính (memoization)
    cache = {1: 1}

    def get_length(k):
        """Hàm đệ quy để tính độ dài chuỗi Collatz."""
        if k not in cache:
            if k % 2 == 0:
                # Nếu k là số chẵn: k -> k / 2
                cache[k] = 1 + get_length(k // 2)
            else:
                # Nếu k là số lẻ: k -> 3k + 1
                cache[k] = 1 + get_length(3 * k + 1)
        return cache[k]
    
    return get_length(n)

def longest_collatz_up_to_n(n):
    """
    Tìm số x trong [1, n] có độ dài chuỗi Collatz lớn nhất
    và trả về x cùng với độ dài L tương ứng.
    Nếu có nhiều số cùng độ dài lớn nhất, chọn số nhỏ nhất.
    """
    if n < 1:
        # Xử lý trường hợp n không phải số nguyên dương
        raise ValueError("Input n must be a positive integer.")

    # Khởi tạo giá trị lớn nhất: Bắt đầu với x=1, L=1
    max_length = 1 
    result_x = 1

    # Duyệt qua tất cả các số từ 1 đến n
    for x in range(1, n + 1):
        current_length = collatz_sequence_length(x)
        
        # Điều kiện:
        # 1. Nếu độ dài hiện tại (current_length) lớn hơn độ dài lớn nhất (max_length) TÌM THẤY
        # 2. **Hoặc** nếu độ dài BẰNG max_length VÀ x hiện tại NHỎ HƠN result_x (Số nhỏ nhất) 
        #    (Điều kiện 2 này là không cần thiết vì chúng ta duyệt từ nhỏ đến lớn, 
        #    nhưng tôi giữ nó ở dạng rõ ràng để phù hợp với logic bài toán: 
        #    Nếu có nhiều số cùng độ dài, chọn số nhỏ nhất. 
        #    Khi duyệt từ 1 đến n, số đầu tiên đạt max_length chính là số nhỏ nhất.)
        
        if current_length > max_length:
            max_length = current_length
            result_x = x
        # Do chúng ta duyệt từ x=1 đến n, nên nếu current_length == max_length 
        # thì result_x đã là số nhỏ nhất có độ dài đó, nên không cần cập nhật.
        # Tuy nhiên, nếu bạn muốn làm code chống lỗi (future-proof), bạn có thể thêm:
        # elif current_length == max_length and x < result_x: 
        #    result_x = x # (Dòng này KHÔNG BAO GIỜ được thực thi khi duyệt từ 1 đến n)
            
    return result_x, max_length

# --- PHẦN CHẠY THỬ CHƯƠNG TRÌNH ---
try:
    # Yêu cầu người dùng nhập số nguyên dương n
    n_input = input("Nhập số nguyên dương n: ")
    n = int(n_input)
    
    if n <= 0:
        print("Lỗi: n phải là một số nguyên dương.")
    else:
        # Gọi hàm chính
        x_result, length_result = longest_collatz_up_to_n(n)
        
        # In kết quả
        print("---")
        print(f"Trong đoạn [1, {n}],")
        print(f"Số có độ dài Collatz lớn nhất là x = **{x_result}**")
        print(f"Với độ dài Collatz tương ứng là L = **{length_result}**")
        print("Ghi chú: Nếu có nhiều số cùng độ dài lớn nhất, chương trình chọn số nhỏ nhất.")

except ValueError:
    print("Lỗi: Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")
```

```python
# --- KHAI BÁO BIẾN ---
# Biến này sẽ lưu trữ độ dài chuỗi Collatz lớn nhất tìm được cho đến nay (L)
max_length = 0 
# Biến này sẽ lưu trữ số nguyên dương có độ dài Collatz lớn nhất (x)
result_x = 0

# --- NHẬP DỮ LIỆU ĐẦU VÀO ---
try:
    # Nhập số nguyên dương n
    n_input = input("Nhập số nguyên dương n: ")
    n = int(n_input)
    
    # Kiểm tra điều kiện n
    if n <= 0:
        print("Lỗi: n phải là một số nguyên dương.")
        # Dừng chương trình nếu n không hợp lệ
        exit() 
        
except ValueError:
    print("Lỗi: Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")
    exit()

# --- VÒNG LẶP NGOÀI: DUYỆT CÁC SỐ TỪ 1 ĐẾN N ---
# Vòng lặp này duyệt qua từng số x trong đoạn [1, n]
for x in range(1, n + 1):
    
    # Bắt đầu tính chuỗi Collatz cho số hiện tại là x
    current_number = x
    # Biến đếm độ dài chuỗi Collatz cho số hiện tại
    current_length = 1 
    
    # --- VÒNG LẶP TRONG: TÍNH ĐỘ DÀI CHUỖI COLLATZ CHO MỘT SỐ ---
    # Chuỗi Collatz kết thúc khi số đó đạt đến 1
    while current_number > 1:
        
        # 1. KIỂM TRA SỐ CHẴN HAY LẺ:
        if current_number % 2 == 0:
            # Nếu là số chẵn: n -> n / 2
            current_number = current_number // 2
        else:
            # Nếu là số lẻ: n -> 3n + 1
            current_number = 3 * current_number + 1
            
        # Tăng độ dài chuỗi lên 1 sau mỗi bước
        current_length = current_length + 1
        
    # --- SO SÁNH KẾT QUẢ VÀ CẬP NHẬT ---
    # So sánh độ dài của chuỗi hiện tại với độ dài lớn nhất đã tìm thấy
    
    if current_length > max_length:
        # Nếu tìm thấy chuỗi dài hơn, cập nhật cả độ dài và số x
        max_length = current_length
        result_x = x
        
    # Ghi chú: Vì chúng ta duyệt x từ 1 đến n (tăng dần),
    # nên nếu có hai số có cùng độ dài lớn nhất, số đầu tiên (nhỏ nhất)
    # đạt được độ dài đó sẽ được giữ lại.

# --- IN KẾT QUẢ CUỐI CÙNG ---
print("\n--- KẾT QUẢ ---")
print(f"Trong đoạn [1, {n}],")
print(f"Số có độ dài Collatz lớn nhất là x = **{result_x}**")
print(f"Với độ dài Collatz tương ứng là L = **{max_length}**")

# Ví dụ minh họa cho n=10:
# x=1, L=1
# x=2, L=2
# x=3, L=8
# x=4, L=3
# x=5, L=6
# x=6, L=9 <--- MAX
# x=7, L=17
# x=8, L=4
# x=9, L=20 <--- MAX MỚI
# x=10, L=7
# Kết quả cuối cùng là x=9, L=20
```

```python
#Hàm đảo ngược các số
def reverse(n):
    n_rv = 0
    while n > 0 :
        n_rv = n_rv*10 + n%10
        n //= 10
    return n_rv
n = int(input('Nhập số:'))
print('-Số bắt đầu: ',n,' -')
print('Nghịch đảo của n: ', reverse(n))
```
