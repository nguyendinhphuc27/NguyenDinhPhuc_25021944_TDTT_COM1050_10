```python
#W4A1
sums = 0
n = int(input('Mời nhập n số nguyên dương đầu cần tính tổng:'))
for i in range(1,n+1):
    sums += i 
print(f'Tổng của {n} số nguyên là: {sums}')
```

```python
#W4A2
n = 0 
c = 0
for _ in range(100000):
    n = int(input('Mời nhập 1 số:'))
    if n >0 :
        for i in range(1,n+1):
            if n % i == 0:
                c+=1
        if c > 2 :
            print(f'{n} không là số nguyên tố')
        else:
                print(f'{n} là số nguyên tố')
        break
    else:
        print('Nhập lại đi!')
```

```python
#w4a4
try:   
    n = int(input("Nhap so n:"))
    n= abs(n)
    count = 0
    temp = n
    for _ in range (100):
        if temp > 0:
            count += 1
            temp //= 10
        else:
            break
    if n == 0:
        count = 1
    print("So chu so la:", count)
except ValueError:
    print("Loi: Ban phai nhap so nguyen!")
```

```python
#W4A6
a, b = map(int, input('Mời nhập khoảng cần kiểm tra:').split())
s = 0
for i in range(a, b+1):
    c=0
    for x in range(1, i+1):
        if i%x==0:
            c=c+1
    if c==2:
        s+=i
print(s)
```

```python
#W4A7
n = int(input('Mời nhập 1 số nguyên dương n (n>=2): '))
lst = []
for i in range (2,n+1):
    if n%i==0 :
        flag = True
        for x in range(2,i):
            if i%x==0:
                flag = False
        if flag == True:
            lst.append(i)
print(lst)
print(f'ước số nguyên tố lớn nhất của {n} là : {max(lst)}')
```
