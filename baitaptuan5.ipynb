```python
#W5A3
def check(x) :
    if x == 2 :
        return True
    for i in range(2,x) :
        if x %i == 0:
            return False
        else:
            return True
while True :
    try:
        x = int(input('Mời nhập số cần kiểm tra: '))
        if x > 0 :
            break
    except ValueError:
        print('Dữ liệu kh hợp lệ! Mời nhập lại: ')
if check(x) == True :
    print(f'{x} là số nguyên tố!')
else:
    print(f'{x} không là số nguyên tố!')
```

```python
#W5A4
def check(x) :
    lst = []
    for i in range(1, x) :
        if x %i == 0:
            lst.append(i)
    if sum(lst) == x :
        return True
while True :
    try:
        x = int(input('Mời nhập số cần kiểm tra: '))
        if x > 0 :
            break
    except ValueError:
        print('Dữ liệu kh hợp lệ! Mời nhập lại: ')
if check(x) == True :
    print(f'{x} là một số hoàn hảo!')
else:
    print(f'{x} kh là số hoàn hảo!')
```

```python
#W5A5
def find_k(k, lst_k) :
    for i in range(len(lst_k)) :
        if lst_k[i] == k :
            return i+1
    else:
        return -1
lst = list(map(int, input('Mời nhập một danh sách của bạn:').split()))
k = int(input('Mời nhập một số k cần kiểm tra: '))
if find_k(k,lst) == -1 :
    print(f'Số {k} không nằm trong danh sách: ',lst)
else:
    print(f'Vị trí của {k} trong danh sách:',lst, ' là',find_k(k,lst))
```

```python
#W5A6
def gt_n(n) :
    gt = 1
    for i in range(1,n+1) :
        gt *= i
    return gt
while True :
    try :
        n = int(input('Mời nhập số cần tính giai thừa: '))
        if n >0 :
            break
        else :
            print('Vui lòng nhập 1 số nguyên dương!')
    except ValueError :
        print('Dữ liệu kh hợp lệ! Mời nhập lại: ')
print(gt_n(n))
```

```python
#W5A7
def calculator(a, b, k ) :
    match k :
        case 'cong' :
            return a + b
        case 'tru' :
            return a - b
        case 'chia' :
            return a / b
        case 'nhan' :
            return a * b
        case _ :
            return False
a, b = map(int, input().split())
k = input('Nhập phép tính cần thực hiện, viết kh dấu: ')
print('Phép ',k,f' với 2 số {a} và {b} là: {calculator(a,b,k):.2f}')
```

```python
#W5A8
def hamming_diff(x, y) :
    a = x ^ y
    return a.bit_count() #đếm số bit 1 trong a 
x, y = map(int, input().split())
print(hamming_diff(x, y))
```

```python
#W5A9
def sum_(n) :
    s = 0
    for i in range(len(str(n))) :
        s += n%10
        n //= 10
    return s
n = int(input('Mời nhập số nguyên dương n:'))
print(sum_(n))
```

```python
def check_isomorphic_strings(a: str, b: str) -> bool:
    """
    Kiểm tra xem hai chuỗi a và b có đẳng cấu (isomorphic) dựa trên quy tắc ánh xạ ký tự hay không.
    Giả định: len(a) == len(b).
    """
    
    # map_a_to_b: Ánh xạ từ ký tự của a sang ký tự của b (dictionary)
    map_a_to_b = {}
    
    # mapped_b_chars: Tập hợp (set) các ký tự của b đã được sử dụng làm đích đến.
    # Dùng để đảm bảo tính ánh xạ một-một (một ký tự của b chỉ được ánh xạ từ duy nhất một ký tự của a).
    mapped_b_chars = set()

    # Hai chuỗi được giả định là có cùng độ dài
    n = len(a)
    
    for i in range(n):
        char_a = a[i]
        char_b = b[i]

        if char_a in map_a_to_b:
            # 1. Kiểm tra tính bảo toàn ánh xạ: 
            # Nếu char_a đã được ánh xạ, nó phải ánh xạ ĐÚNG char_b hiện tại.
            if map_a_to_b[char_a] != char_b:
                return False
        else:
            # 2. char_a chưa được ánh xạ (lần xuất hiện đầu tiên)
            
            # Kiểm tra tính một-một: char_b chưa được sử dụng làm đích đến 
            # bởi một ký tự khác của a.
            if char_b in mapped_b_chars:
                # Nếu char_b đã được dùng, tức là có 2 ký tự khác nhau của a 
                # ánh xạ đến cùng 1 ký tự của b. => VI PHẠM
                return False
            
            # Thiết lập ánh xạ mới
            map_a_to_b[char_a] = char_b
            
            # Đánh dấu char_b là đã được sử dụng
            mapped_b_chars.add(char_b)

    # Nếu vòng lặp hoàn thành mà không có vi phạm, hai chuỗi là đẳng cấu.
    return True

# --- Ví dụ kiểm tra ---

# Ví dụ trong đề bài: "abca" và "zbxz"
print(f"'abca' và 'zbxz': {check_isomorphic_strings('abca', 'zbxz')}") 
# Ánh xạ: a -> z, b -> b, c -> x. Đúng. -> True

# Ví dụ kiểm tra tính một-một: "foo" và "bar"
# Ánh xạ: f -> b, o -> a, o -> r. 'o' cần ánh xạ đến 'a' và 'r' => VI PHẠM
print(f"'foo' và 'bar': {check_isomorphic_strings('foo', 'bar')}") 
# Logic: f->b, o->a. Đến ký tự thứ 3: a[2]='o', b[2]='r'.
# Nhưng 'o' đã được ánh xạ thành 'a'. map_a_to_b['o'] != 'r'. -> False

# Ví dụ kiểm tra tính một-một: "egg" và "add"
# Ánh xạ: e -> a, g -> d. Đúng.
print(f"'egg' và 'add': {check_isomorphic_strings('egg', 'add')}")
# Logic: e->a. g->d. Ký tự thứ 3: a[2]='g', b[2]='d'. 
# map_a_to_b['g'] == 'd'. -> True

# Ví dụ vi phạm tính một-một (hai ký tự a khác nhau cùng ánh xạ đến một ký tự b)
# "ab" và "aa"
# Ánh xạ: a -> a, b -> a. 'a' và 'b' khác nhau nhưng cùng ánh xạ đến 'a' => VI PHẠM
print(f"'ab' và 'aa': {check_isomorphic_strings('ab', 'aa')}")
# Logic: a->a. Đến ký tự thứ 2: a[1]='b', b[1]='a'. 
# 'a' đã có trong mapped_b_chars. 'b' chưa có trong map_a_to_b. -> False

# Ví dụ kiểm tra: "paper" và "title"
# Ánh xạ: p->t, a->i, p->t (OK), e->l, r->e.
print(f"'paper' và 'title': {check_isomorphic_strings('paper', 'title')}") 
# Logic: p->t, a->i. Đến 'p' tiếp theo: p->t (OK). e->l. r->e. -> True
```
