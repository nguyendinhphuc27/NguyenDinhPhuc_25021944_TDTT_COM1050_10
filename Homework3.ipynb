```python
#HW3.4
while True:
    h = int(input('Mời nhập chiều cao của tam giác: '))
    if h > 0 :
        break
    continue
for i in range(1, h+1) :
    if i == h :
       print('*'*(2*h-1))
    elif i == 1 :
        space = h -1
        print(' '*space + '*')
    else :
        space = h - i
        sp = 2*i - 3  # số * ở dòng i: 2*(i-1) + 1 => khoảng trống trừ 2 * ở đầu = 2i-3
        print(' '*space + '*' + ' '*sp + '*')
```

```python
#HW3.7
a = 0
while True:
    try: 
        a = int(input('Mời nhập số dương a:'))
        if a > 0 :
            break
        continue
    except :
        print('Không hợp lệ!')
count, c = 0, 0
n = 1
for i in range(1, a+1):
    if a %i == 0:
        count += 1
print(count)
while n <= a:
    if a %n == 0:
        c += 1
    n += 1
print(c)
```

```python
#HHW3.8
a, b  = 0, 0
while True :
    a, b = map(int, input('Mời nhập hai số nguyên dương cần kiểm tra:').split())
    if a > 0 and b > 0 :
        break
    continue
min_ = min(a,b)
u = 1
for i in range(1,min_+1):
    if min_ %i == 0 and max(a,b) %i == 0 :
        print(i, end = ' ')
print(' ', end ='\n' )
while u <= min_ :
    if min_ %u == 0 and max(a,b) %u == 0 :
        print(u, end = ' ')
    u += 1
```

```python
#HW3.9
import math
a = 0
while True :
    a = int(input('Moi nhap so duong a can kiem tra:'))
    if a > 0 :
        break
    continue
i = 1
while i < int(math.sqrt(a)+1) :
    if i**2 == a :
        print(f'{a} la mot so chinh phuong')
        break
    i += 1
    if i == int(math.sqrt(a)) and i**2 != a :
        print(f'{a} khong la mot so chinh phuong')
```
