```python
#HW3
''' 
  _hàm kiểm tra ký tự hoa và ký tự thường_
''biến'.isupper()': kiểm tra xem có phải kí tự hoa kh
''biến'.islower()': kiểm tra xem có phải kí tự thường kh
''biến'.lower()':chuyển từ ký tự hoa sang ký tự thường
''biến'.upper()':chuyển từ ký tự thg sang ký tự hoa
'''
a= input('Mời bạn nhập 1 ký tự:')
if a.isupper()== True:
    print(a.lower())
else:
    print(a.upper())
```

```python
#HW8
a, b = map(int,input('số điện tháng trc và tháng này của hộ:').split())
n= input('Tên chủ hộ là:')
if (b-a)<51:
    s=(b-a)*1984
    print('Họ và tên: '+n,'Tiền phải trả là: '+f'{s*1.08:.0f}', sep='\n')
elif 50<(b-a)<101:
    s=50*1984  +(b-a-50)*2050
    print('Họ và tên: '+n,'Tiền phải trả là: '+f'{s*1.08:.0f}', sep='\n')
elif 100<(b-a)<201:
    s= 50*1984 + 50*2050 + (b-a-100)*2380
    print('Họ và tên: '+n,'Tiền phải trả là: '+f'{s*1.08:.0f}', sep='\n')
elif 200<b-a<301:
    s=50*1984 + 50*2050 + 100*2380 + (b-a-200)*2998
    print('Họ và tên: '+n,'Tiền phải trả là: '+f'{s*1.08:.0f}', sep='\n')
elif 300<b-a<401:
    s=50*(1984+2050) + 100*(2380+2998)+ (b-a-300)*3350
    print('Họ và tên: '+n,'Tiền phải trả là: '+f'{s*1.08:.0f}', sep='\n')
else:
    s=50*(1984+2050) + 100*(2380+2998+3350)+(b-a-400)*3460
    print('Họ và tên: '+n,'Tiền phải trả là: '+f'{s*1.08:.0f}', sep='\n')
```
